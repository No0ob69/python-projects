```python
import random
guess = input("guess the number. from 1-10: ")
sec_num= random.randint(1, 10)
tries =1
guess =int(guess)
while sec_num != guess:
  if guess> 10 or guess < 1:
    print('invalid guess try again')
  else:
     print("try again ")
  guess = int(input("guess the number from 1-10: "))
  tries += 1
if sec_num == guess:
   print(f"you got it in {tries} guesses!")
```
