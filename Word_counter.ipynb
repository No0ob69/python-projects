```python
paragraph = input("Add your paragraph here: ")
target = input("Add the word you want to count here: ")
count = 0

words = paragraph.lower().split()
for word in words:
    if word == target.lower():
            count += 1

print(f"{target} appears {count} times")
```

```python

paragraph = (input('add your text here:'))
count= {}

words = paragraph.lower().split()
for word in words:
  if word in count:
    count[word] += 1
  else:
    count[word] = 1

for key, value in count.items:
  print(f'{key} appears {value} times' )
```
