```python
#being able to add names and conacts, delete them and search them
contacts={}

def adding_contact(name, number):
  contacts[name] = number
def deleting_contact(name):
  if name in contacts:
    del contacts[name]
  else:
    print('contact name doesnt exist')
def view_all() :
  for name, number in contacts.items():
    print(f'contact info for {name} is {number} ')
def search(name):
  print(contacts.get(name, 'contact not found'))

while True:
  choice= input("search, add, delete, view all or quit").lower()
  if choice== 'add':
    name =input('contact name: ')
    number = input('contact number:')
    adding_contact(name,number)
  elif choice == 'search':
    name= input('what is the name of the contact: ')
    search(name)
  elif choice== 'delete':
    name= input("name of the contact you want to delete: ")
    deleting_contact(name)
  elif choice == 'view all':
    view_all()
  elif choice== 'quit':
    break
  else:
    print ('please enter a valid command')






```
